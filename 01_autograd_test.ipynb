```python
import matplotlib.pyplot as plt
from nn.model import MLP
from nn.optim import SGD
from nn.losses import MSELoss
```

```python
criterion = MSELoss()

nn = MLP(3, [4, 4, 1])
optimizer = SGD(nn.parameters(), lr=0.05)

xs = [[2.0, 3.0, -1.0], [3.0, -1.0, 0.5], [0.5, 1.0, 1.0], [-1.0, -0.5, -1.0]]
ys = [1.0, -1.0, -1.0, 1.0]

loss_history = []

epochs = 30
for k in range(epochs):
    ypred = [nn(x) for x in xs]
    loss = criterion(ys,ypred)
    
    loss_history.append(loss.data)
    
    optimizer.zero_grad() 
    loss.backward()
    optimizer.step()

plt.figure(figsize=(8, 5))
plt.plot(loss_history, color='cyan', marker='o')
print(f"Final Loss at Epoch {epochs-1}: {loss_history[-1]:.4f}")
plt.title('Framework Loss Drop Over Time')
plt.xlabel('Epoch')
plt.ylabel('Loss Value')
plt.grid(True, linestyle='--', alpha=0.6)
plt.style.use('dark_background')
plt.show()
```
