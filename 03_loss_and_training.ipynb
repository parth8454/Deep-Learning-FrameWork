```python
nn = MLP(3,[4,4,1])

xs = [
    [2,3,-1],[3,-1,0.5],[0.5,1,1],[-1,-0.5,-1]
]

ys = [1.0,-1.0,-1.0,1.0]

ypred = [nn(x) for x in xs]

for p in ypred:
    print(p.data)
```

```python
def mean_squared_error(y_true,y_pred):
    losses = [(yout - ygt)**2 for ygt,yout in zip(y_true,y_pred)]
    total_loss = losses[0]
    for l in losses[1:]:
        total_loss += l
    return total_loss*(1.0/len(y_true))

epochs = 20
lr = 0.05
xs = [
    [2,3,-1],[3,-1,0.5],[0.5,1,1],[-1,-0.5,-1]
]
for k in range(epochs):

    ypred = [nn(x) for x in xs]

    loss = mean_squared_error(ys,ypred)

    for p in nn.parameters():
        p.grad = 0.0

    loss.backward()

    print(f"Loss gradient: {loss.grad}")
    print(f"First parameter gradient: {nn.parameters()[0].grad}")

    for p in nn.parameters():
        p.data -= lr*p.grad

    print("\nfinal predic...")
    for p in ypred:
        print(f"{p.data:.4f}")
```

```python
from nn.model import MLP

# 1. FORCE a brand new network to be created with the NEW code!
nn = MLP(3, [4, 4, 1])

# 2. Re-define the data just in case
xs = [
    [2.0, 3.0, -1.0], 
    [3.0, -1.0, 0.5], 
    [0.5, 1.0, 1.0], 
    [-1.0, -0.5, -1.0]
]
ys = [1.0, -1.0, -1.0, 1.0]

# 3. The exact same training loop
epochs = 20
lr = 0.05

for k in range(epochs):
    # Forward
    ypred = [nn(x) for x in xs]
    loss = mean_squared_error(ys, ypred)
    
    # Zero grad
    for p in nn.parameters():
        p.grad = 0.0
        
    # Backward
    loss.backward()
    
    # Update
    for p in nn.parameters():
        p.data -= lr * p.grad
        
    print(f"Epoch {k} | Loss: {loss.data:.4f}")

# Final predictions
print("\nFinal Predictions:")
for p in ypred:
    print(f"{p.data:.4f}")
```
